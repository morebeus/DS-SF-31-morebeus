```python
import smtplib
import time
import imaplib
import email
import traceback
import pandas as pd

# -------------------------------------------------
# Utility to read email from Gmail Using Python
# ------------------------------------------------
ORG_EMAIL = "@gmail.com"
FROM_EMAIL = "jo9435675" + ORG_EMAIL
FROM_PWD = "wkbf vbpu ayzh eztx"  # Replace with your app password
SMTP_SERVER = "imap.gmail.com"
SMTP_PORT = 993

all_messages = pd.DataFrame()

def read_email_from_gmail():
    all_msg = pd.DataFrame()
    email_calls = []
    try:
        mail = imaplib.IMAP4_SSL(SMTP_SERVER)
        mail.login(FROM_EMAIL, FROM_PWD)
        mail.select('inbox')
        data = mail.search(None, 'ALL')
        mail_ids = data[1]
        id_list = mail_ids[0].split()
        first_email_id = int(id_list[0])
        latest_email_id = int(id_list[-1])

        for i in range(latest_email_id, first_email_id, -1):
            print("<< Message ", i, ">>")
            data = mail.fetch(str(i), '(RFC822)')
            all_messages[i] = data
            for response_part in data:
                arr = response_part[0]
                if isinstance(arr, tuple):
                    msg = email.message_from_bytes(arr[1])  # Use message_from_bytes
                    email_subject = msg['subject']
                    email_from = msg['from']
                    email_to = msg['Delivered-To']
                    email_date = msg['date']
                    if msg['reply-to'] is None:
                        email_return = ""
                    else:
                        email_return = msg['reply-to']

                    print('From : ' + email_from + '\n')
                    print('To : ' + email_to + '\n')
                    print('Subject : ' + email_subject + '\n')
                    print('Date : ' + email_date + '\n')
                    print('email return : ' + email_return + '\n')
                    email_calls.append(msg.keys())
    except Exception as e:
        traceback.print_exc()
        print(str(e))
    return email_calls, all_messages


calls_list, messages = read_email_from_gmail()
```
